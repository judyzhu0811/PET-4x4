```python
#importing
import csv
import numpy as np
import matplotlib.pyplot as plt
import copy
import math
import keras
import pandas as pd
import xgboost as xgb
import os

from keras.models import Sequential
from keras.layers import Dense, BatchNormalization, Dropout
from keras.callbacks import EarlyStopping
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error
from sklearn.metrics import roc_auc_score
```

```python
numbers = [3.05, 4.05, 5.05, 6.05, 7.05, 8.05, 9.05, 10.05, 11.05, 12.05, 13.05, 14.05, 14.3, 14.55, 14.8, 15.05, 15.3, 15.55, 15.8, 16.05, 16.3, 16.55, 16.8, 17.05, 17.3, 17.55, 17.8, 18.05, 18.3, 18.55, 18.8, 19.05, 20.05, 25.05, 40.05]

sipm_coordinates = [
    [-24.375, -24.375],
    [-24.375,  -8.125],
    [-24.375,   8.125],
    [-24.375,  24.375],

    [ -8.125, -24.375],
    [ -8.125,  -8.125],
    [ -8.125,   8.125],
    [ -8.125,  24.375],

    [  8.125, -24.375],
    [  8.125,  -8.125],
    [  8.125,   8.125],
    [  8.125,  24.375],

    [ 24.375, -24.375],
    [ 24.375,  -8.125],
    [ 24.375,   8.125],
    [ 24.375,  24.375]
]
closest_distances_all_data = []
weighted_distances_all_data = []
XGBoost_distances_all_data = []

for q in numbers:
    file_path = os.path.join('/Users/judyz/Desktop/PET-4x4/build',f'{q}mm',f'{q}mm.csv')
    file = open(file_path)
    csvreader = csv.reader(file)

    line_count = 0

    original_X = []
    original_y = []

    for row in csvreader:
        if line_count >= 1:
            original_X.append([int(i) for i in row[1:17]])
            original_y.append([float(row[17]),float(row[18])])
        line_count += 1

    file.close()

    X = np.array(original_X)
    y = np.array(original_y)

    print("Number of events:", len(X))
    print("X shape:", X.shape)
    print("y shape:", y.shape)
    print("X coordinate range:",np.min(y[:, 0]),"to",np.max(y[:, 0]))
    print("Y coordinate range:",np.min(y[:, 1]),"to",np.max(y[:, 1]))

    # 331 x 331 grid; 0.2 mm spacing; center is index 165
    closest_distances = np.zeros((331, 331))

    for i in range(len(X)):
        maximum_value = np.argmax(X[i])
        reconstructed_position = sipm_coordinates[maximum_value]

        x_index = 165 + round(y[i][0] * 5)
        y_index = 165 - round(y[i][1] * 5)

        if 0 <= x_index < 331 and 0 <= y_index < 331:
            closest_distances[x_index][y_index] = math.dist(reconstructed_position,y[i])
    print("Average Regular Distance (mm):",np.average(closest_distances))
    print("Median Regular Distance (mm):",np.median(closest_distances))

    #weighted average
    weighted_distances = np.zeros((331, 331))
    for i in range(len(X)):
        tmp = X[i]
        total_signal = sum(tmp)

        if total_signal == 0:
            continue

        coord1 = sum(
            tmp[j] * sipm_coordinates[j][0]
            for j in range(16)
        ) / total_signal

        coord2 = sum(
            tmp[j] * sipm_coordinates[j][1]
            for j in range(16)
        ) / total_signal


        x_index = 165 + round(y[i][0] * 5)
        y_index = 165 - round(y[i][1] * 5)

        if 0 <= x_index < 331 and 0 <= y_index < 331:
            weighted_distances[x_index][y_index] = math.dist([coord1, coord2],y[i])

    print("Average Weighted Distance (mm):",np.average(weighted_distances))
    print("Median Weighted Distance (mm):",np.median(weighted_distances))

    #xgboost
    XGBoost_distances = np.zeros((331, 331))
    scaler = StandardScaler()

    for k in range(10):
        X_XGBoost = scaler.fit_transform(X)
        X_train, X_val, y_train, y_val = train_test_split(X_XGBoost,y,test_size=0.3)

        model = xgb.XGBRegressor()
        model.fit(X_train,y_train)

        yhat = model.predict(X_XGBoost)

        for i in range(len(X)):

            x_index = 165 + round(y[i][0] * 5)
            y_index = 165 - round(y[i][1] * 5)

            if 0 <= x_index < 331 and 0 <= y_index < 331:

                distance = math.dist(yhat[i],y[i])
            
                XGBoost_distances[x_index][y_index] = (k / (k + 1) * XGBoost_distances[x_index][y_index] + 1 / (k + 1)* distance)
     
        weighted_nonzero = weighted_distances[weighted_distances > 0]
        closest_nonzero = closest_distances[closest_distances > 0]
        XGBoost_nonzero = XGBoost_distances[XGBoost_distances > 0]
        print("Average Weighted Distance (mm):", np.average(weighted_nonzero))
        print("Median Weighted Distance (mm):",np.median(weighted_nonzero))
        print("Average Regular Distance (mm):",np.average(closest_nonzero))
        print("Median Regular Distance (mm):",np.median(closest_nonzero))
        print("Average XGBoost Distance (mm):",np.average(XGBoost_nonzero))
        print("Median XGBoost Distance (mm):",np.median(XGBoost_nonzero))
    
    plt.figure()
    plt.hist(closest_distances[closest_distances > 0],bins=100,alpha=0.6,label='Regular')
    plt.hist(weighted_distances[weighted_distances > 0],bins=100,alpha=0.6,label='Weighted Average')
    plt.hist(XGBoost_distances[XGBoost_distances > 0],bins=100,alpha=0.6,label='XGBoost')
    plt.xlabel('Distance Between Original Point and Reconstructed Point (mm)')
    plt.ylabel('Counts')
    plt.legend(loc='upper right')
    plt.show()

    xaxis = np.linspace(-33, 33, 331)
    yaxis = np.linspace(-33, 33, 331)

    plt.figure(figsize=(8, 6))
    plt.pcolormesh(xaxis,yaxis,XGBoost_distances,shading='auto',cmap='viridis')
    plt.colorbar(label='Distance Between Original and Reconstructed Point (mm)')
    plt.xlabel('X (mm)')
    plt.ylabel('Y (mm)')
    plt.title('Accuracy of XGBoost')
    plt.show()


    #weighted average
    plt.figure(figsize=(8, 6))
    plt.pcolormesh(xaxis,yaxis,weighted_distances,shading='auto',cmap='viridis')
    plt.colorbar(label='Distance Between Original and Reconstructed Point (mm)')
    plt.xlabel('X (mm)')
    plt.ylabel('Y (mm)')
    plt.title('Accuracy of Weighted Average')
    plt.show()

    #closest average
    plt.figure(figsize=(8, 6))
    plt.pcolormesh(xaxis,yaxis,closest_distances,shading='auto',cmap='viridis')
    plt.colorbar(label='Distance Between Original and Reconstructed Point (mm)')
    plt.xlabel('X (mm)')
    plt.ylabel('Y (mm)')
    plt.title('Accuracy of Regular Method')
    plt.show()


    closest_distances_all_data.append(closest_distances)
    weighted_distances_all_data.append(weighted_distances)
    XGBoost_distances_all_data.append(XGBoost_distances)
```

```python
closest_distances_all_data_np = np.array(closest_distances_all_data)
weighted_distances_all_data_np = np.array(weighted_distances_all_data)
XGBoost_distances_all_data_np = np.array(XGBoost_distances_all_data)

np.savetxt('closest_distances_xgboost.csv', closest_distances_all_data_np.reshape(closest_distances_all_data_np.shape[0], -1))
np.savetxt('weighted_distances_xgboost.csv', weighted_distances_all_data_np.reshape(weighted_distances_all_data_np.shape[0], -1))
np.savetxt('XGBoost_distances_xgboost.csv', XGBoost_distances_all_data_np.reshape(XGBoost_distances_all_data_np.shape[0], -1))

print(XGBoost_distances_all_data_np.shape)
```
