```python
import csv
import numpy as np
import matplotlib.pyplot as plt
import copy
import math
import keras
import pandas as pd
import os

from keras.models import Sequential
from keras.layers import Dense, BatchNormalization, Dropout
from keras.callbacks import EarlyStopping
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error
```

```python
MLP_distances_all_data = []
```

```python
numbers = [3.05, 4.05, 5.05, 6.05, 7.05,8.05, 9.05, 10.05, 11.05, 12.05,13.05, 14.05, 15.05, 16.05, 17.05, 18.05, 19.05, 20.05, 25.05, 40.05]

# x = -30 mm to 30 mm; y = -30 mm to 30 mm

region_size = 30
grid_half_size = round(region_size * 5)  
grid_size = 2 * grid_half_size + 1       

for q_num, q in enumerate(numbers, start=1):

    file_path = os.path.join('/Users/judyz/Desktop/PET-4x4/build',f'{q}mm',f'{q}mm.csv')
    csvreader = csv.reader(open(file_path))

    original_X = []
    original_y = []

    line_count = 0
    for row in csvreader:
        if line_count >= 1:
            original_X.append([int(i) for i in row[1:17]])
            original_y.append([float(row[17]),float(row[18])])
        line_count += 1

    X = np.array(original_X)
    y = np.array(original_y)

    print("Original number of events:", len(X))
    print("Original X shape:", X.shape)
    print("Original y shape:", y.shape)

    print(
        "Original X coordinate range:",
        np.min(y[:, 0]),
        "to",
        np.max(y[:, 0])
    )

    print(
        "Original Y coordinate range:",
        np.min(y[:, 1]),
        "to",
        np.max(y[:, 1])
    )


    mask = (
        (y[:, 0] >= -region_size) &
        (y[:, 0] <= region_size) &
        (y[:, 1] >= -region_size) &
        (y[:, 1] <= region_size)
    )

    X = X[mask]
    y = y[mask]


    print("\nAfter 5 mm cut:")
    print("Number of events:", len(X))
    print("X shape:", X.shape)
    print("y shape:", y.shape)

    print(
        "X coordinate range:",
        np.min(y[:, 0]),
        "to",
        np.max(y[:, 0])
    )

    print(
        "Y coordinate range:",
        np.min(y[:, 1]),
        "to",
        np.max(y[:, 1])
    )

    MLP_distances = np.zeros(
        (grid_size, grid_size, 2)
    )

    scaler = StandardScaler()

    for k in range(10):

        print(
            f"Distance {q_num}/{len(numbers)} "
            f"({q} mm) | MLP run {k+1}/10"
        )

        X_MLP = scaler.fit_transform(X)
        X_train, X_val, y_train, y_val = train_test_split(X_MLP, y,test_size=0.3)
    
        model = Sequential()
        model.add(
            Dense(
                128,
                input_dim=16,
                activation='relu'
            )
        )
        model.add(BatchNormalization())
        model.add(Dropout(0.2))
        model.add(
            Dense(
                128,
                activation='relu'
            )
        )
        model.add(BatchNormalization())
        model.add(Dropout(0.2))
        model.add(
            Dense(
                128,
                activation='relu'
            )
        )
        model.add(BatchNormalization())
        model.add(Dropout(0.2))
        model.add(
            Dense(
                128,
                activation='relu'
            )
        )
        model.add(BatchNormalization())
        model.add(Dropout(0.2))
        model.add(
            Dense(
                2,
                activation='linear'
            )
        )

        model.compile(
            optimizer='adam',
            loss='mean_squared_error'
        )

        history = model.fit(
            X_train,
            y_train,
            epochs=100,
            batch_size=32,
            verbose=0
        )

        yhat = model.predict(X_MLP,verbose=0)
            
        for i in range(len(X)):

            x_index = (
                grid_half_size
                +
                round(y[i][0] * 5)
            )

            y_index = (
                grid_half_size
                -
                round(y[i][1] * 5)
            )


            if (
                0 <= x_index < grid_size
                and
                0 <= y_index < grid_size
            ):

                MLP_distances[
                    x_index,
                    y_index,
                    0
                ] = (
                    k / (k + 1)
                    *
                    MLP_distances[
                        x_index,
                        y_index,
                        0
                    ]
                    +
                    1 / (k + 1)
                    *
                    (
                        yhat[i][0]
                        -
                        y[i][0]
                    )
                )
                MLP_distances[
                    x_index,
                    y_index,
                    1
                ] = (
                    k / (k + 1)
                    *
                    MLP_distances[
                        x_index,
                        y_index,
                        1
                    ]
                    +
                    1 / (k + 1)
                    *
                    (
                        yhat[i][1]
                        -
                        y[i][1]
                    )
                )

    delta_x = MLP_distances[:, :, 0].ravel()
    delta_y = MLP_distances[:, :, 1].ravel()
    delta_x_nonzero = delta_x[delta_x != 0]
    delta_y_nonzero = delta_y[delta_y != 0]

    print("\nMLP Results")

    print(
        "Average MLP Delta X (mm):",
        np.mean(delta_x_nonzero)
    )

    print(
        "Median MLP Delta X (mm):",
        np.median(delta_x_nonzero)
    )

    print(
        "Average MLP Delta Y (mm):",
        np.mean(delta_y_nonzero)
    )

    print(
        "Median MLP Delta Y (mm):",
        np.median(delta_y_nonzero)
    )

    plt.figure()

    plt.hist(
        delta_x_nonzero,
        bins=100,
        alpha=0.6,
        label='delta X'
    )

    plt.hist(
        delta_y_nonzero,
        bins=100,
        alpha=0.6,
        label='delta Y'
    )

    plt.xlabel(
        'Distance Between Original Point and Reconstructed Point (mm)'
    )

    plt.ylabel('Counts')

    plt.legend(loc='upper right')

    plt.title(
        f'MLP Reconstruction Error — SiPM Distance {q} mm'
    )

    plt.show()

    MLP_distances_all_data.append(
        MLP_distances
    )

MLP_distances_all_data_np = np.array(
    MLP_distances_all_data
)

print(
    "\nFinal array shape:",
    MLP_distances_all_data_np.shape
)
```
