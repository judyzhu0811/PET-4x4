```python
#importing
import csv
import numpy as np
import matplotlib.pyplot as plt
import copy
import math
import keras
import pandas as pd
import os

from keras.models import Sequential
from keras.layers import Dense, BatchNormalization, Dropout
from keras.callbacks import EarlyStopping
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error
```

```python
closest_distances_all_data = []
weighted_distances_all_data = []
MLP_distances_all_data = []
```

```python
numbers = [3.05, 4.05, 5.05, 6.05, 7.05,8.05, 9.05, 10.05, 11.05, 12.05,13.05, 14.05, 15.05, 16.05, 17.05, 18.05, 19.05, 20.05, 25.05, 40.05]

sipm_coordinates = [
    [-24.375, -24.375],
    [-24.375,  -8.125],
    [-24.375,   8.125],
    [-24.375,  24.375],

    [ -8.125, -24.375],
    [ -8.125,  -8.125],
    [ -8.125,   8.125],
    [ -8.125,  24.375],

    [  8.125, -24.375],
    [  8.125,  -8.125],
    [  8.125,   8.125],
    [  8.125,  24.375],

    [ 24.375, -24.375],
    [ 24.375,  -8.125],
    [ 24.375,   8.125],
    [ 24.375,  24.375]
]

for q in numbers:
    file_path = os.path.join('/Users/judyz/Desktop/PET-4x4/build',f'{q}mm',f'{q}mm.csv')
    file = open(file_path)
    csvreader = csv.reader(file)

    line_count = 0

    original_X = []
    original_y = []

    for row in csvreader:

        if line_count >= 1:

            original_X.append([int(i) for i in row[1:17]])
            original_y.append([float(row[17]),float(row[18])])
        line_count += 1
    file.close()

    X = np.array(original_X)
    y = np.array(original_y)

    print("Number of events:", len(X))
    print("X shape:", X.shape)
    print("y shape:", y.shape)
    print("X coordinate range:",np.min(y[:, 0]),"to",np.max(y[:, 0]))
    print("Y coordinate range:",np.min(y[:, 1]), "to", np.max(y[:,1]))

    # 331 x 331 grid; Center = index 165
    closest_distances = np.zeros((331, 331))
    for i in range(len(X)):

        maximum_value = np.argmax(X[i])
        reconstructed_position = sipm_coordinates[maximum_value]

        x_index = 165 + round(y[i][0] * 5)
        y_index = 165 - round(y[i][1] * 5)

        if 0 <= x_index < 331 and 0 <= y_index < 331:
            closest_distances[x_index][y_index] = math.dist(
                reconstructed_position,
                y[i]
            )

    closest_nonzero = closest_distances[closest_distances > 0]
    print("Average Regular Distance (mm):",np.average(closest_nonzero))
    print("Median Regular Distance (mm):",np.median(closest_nonzero))

    weighted_distances = np.zeros((331, 331))
    for i in range(len(X)):
        tmp = X[i]
        total_signal = np.sum(tmp)

        if total_signal == 0:
            continue

        coord1 = sum(
            tmp[j] * sipm_coordinates[j][0]
            for j in range(16)
        ) / total_signal

      
        coord2 = sum(
            tmp[j] * sipm_coordinates[j][1]
            for j in range(16)
        ) / total_signal

        x_index = 165 + round(y[i][0] * 5)
        y_index = 165 - round(y[i][1] * 5)

        if 0 <= x_index < 331 and 0 <= y_index < 331:
            weighted_distances[x_index][y_index] = math.dist([coord1, coord2],y[i])
 
    weighted_nonzero = weighted_distances[weighted_distances > 0]

    print("Average Weighted Distance (mm):",np.average(weighted_nonzero))
    print("Median Weighted Distance (mm):",np.median(weighted_nonzero))

    MLP_distances = np.zeros((331, 331))
    scaler = StandardScaler()

    for k in range(10):
        print("Training MLP:", k + 1, "/ 10")

        X_MLP = scaler.fit_transform(X)
        X_train, X_val, y_train, y_val = train_test_split(X_MLP,y, test_size=0.3)

        model = Sequential()

        model.add(Dense(128,input_dim=16,activation='relu'))
        model.add(BatchNormalization())
        model.add(Dropout(0.2))
        model.add(Dense(128,activation='relu'))
        model.add(BatchNormalization())
        model.add(Dropout(0.2))
        model.add(Dense(128,activation='relu'))
        model.add(BatchNormalization())
        model.add(Dropout(0.2))
        model.add(Dense(128,activation='relu'))
        model.add(BatchNormalization())
        model.add(Dropout(0.2))
        model.add(Dense(2,activation='linear'))
   
        model.compile(optimizer='adam',loss='mean_squared_error')

        history = model.fit(X_train,y_train,epochs=100,batch_size=32,verbose=0)
        yhat = model.predict(X_MLP,verbose=0)

        for i in range(len(X)):

            x_index = 165 + round(y[i][0] * 5)
            y_index = 165 - round(y[i][1] * 5)

            if 0 <= x_index < 331 and 0 <= y_index < 331:

                distance = math.dist(yhat[i],y[i])
          
                MLP_distances[x_index][y_index] = (k/(k+1)*MLP_distances[x_index][y_index]+1/(k+1)*distance)

    MLP_nonzero = MLP_distances[MLP_distances > 0]

    print("\nFINAL RESULTS FOR", q, "mm")
    print("Average Regular Distance (mm):",np.average(closest_nonzero))
    print("Median Regular Distance (mm):",np.median(closest_nonzero))
    print("Average Weighted Distance (mm):",np.average(weighted_nonzero))
    print("Median Weighted Distance (mm):",np.median(weighted_nonzero))
    print("Average MLP Distance (mm):",np.average(MLP_nonzero))
    print("Median MLP Distance (mm):",np.median(MLP_nonzero))

    plt.figure(figsize=(8, 6))
    plt.hist(closest_nonzero,bins=100,alpha=0.6,label='Regular')
    plt.hist(weighted_nonzero,bins=100,alpha=0.6,label='Weighted Average')
    plt.hist(MLP_nonzero,bins=100,alpha=0.6,label='MLP')
    plt.xlabel('Distance Between Original Point and Reconstructed Point (mm)')
    plt.ylabel('Counts')
    plt.title(f'Reconstruction Accuracy - {q} mm')
    plt.legend(loc='upper right')
    plt.show()

    xaxis = np.linspace(-33, 33, 331)
    yaxis = np.linspace(-33, 33, 331)

    # MLP Map
    plt.figure(figsize=(8, 6))
    plt.pcolormesh(xaxis,yaxis,MLP_distances,shading='auto',cmap='viridis')
    plt.colorbar(label='Distance Between Original and Reconstructed Point (mm)')
    plt.xlabel('X (mm)')
    plt.ylabel('Y (mm)')
    plt.title(f'Accuracy of MLP - {q} mm')
    plt.show()


    # weighted average
    plt.figure(figsize=(8, 6))
    plt.pcolormesh(xaxis,yaxis,weighted_distances,shading='auto',cmap='viridis')
    plt.colorbar(label='Distance Between Original and Reconstructed Point (mm)')
    plt.xlabel('X (mm)')
    plt.ylabel('Y (mm)')
    plt.title(f'Accuracy of Weighted Average - {q} mm')
    plt.show()

    # regular method
    plt.figure(figsize=(8, 6))
    plt.pcolormesh(xaxis,yaxis,closest_distances,shading='auto',cmap='viridis')
    plt.colorbar(label='Distance Between Original and Reconstructed Point (mm)')
    plt.xlabel('X (mm)')
    plt.ylabel('Y (mm)')
    plt.title(f'Accuracy of Regular Method - {q} mm')
    plt.show()

    closest_distances_all_data.append(closest_distances)
    weighted_distances_all_data.append(weighted_distances)
    MLP_distances_all_data.append(MLP_distances)
```

```python
closest_distances_all_data_np = np.array(closest_distances_all_data)
weighted_distances_all_data_np = np.array(weighted_distances_all_data)
MLP_distances_all_data_np = np.array(MLP_distances_all_data)

np.savetxt('closest_distances.csv', closest_distances_all_data_np.reshape(closest_distances_all_data_np.shape[0], -1))
np.savetxt('weighted_distances.csv', weighted_distances_all_data_np.reshape(weighted_distances_all_data_np.shape[0], -1))
np.savetxt('MLP_distances.csv', MLP_distances_all_data_np.reshape(MLP_distances_all_data_np.shape[0], -1))

print(MLP_distances_all_data_np.shape)
```
